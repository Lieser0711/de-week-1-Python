````python
def unique_words(s: str):
    set_str = set(s.lower().split())
    return(set_str)

assert unique_words("Red red BLUE blue") == {"red","blue"}
````

````python
def is_pangram(s: str):
    test_s = s.replace(" ", "").lower()
    if not test_s.isalnum():
        return False
    else:
        set_str = set(test_s)
        if len(set_str) == 26:
            return True
        else:
            return False
        
assert is_pangram("The quick brown fox jumps over a lazy dog")
````

````python
def disjoint(a_tags, b_tags):
    intersection = a_tags.intersection(b_tags) 
    if len(intersection) > 0:
        return False
    else:
        return True

assert disjoint({"sql","python"}, {"go"}) is True
assert disjoint({"sql","python"}, {"python","go"}) is False
````

````python
def set_ops(a, b):
    """Return dict with union, intersection, diff_a_b, diff_b_a, symdiff."""
    d = dict()
    d["union"] = a.union(b)
    d["intersection"] = a.intersection(b)
    d["diff_a_b"] = a.difference(b)
    d["diff_b_a"] = b.difference(a)
    d["symdiff"] = a.symmetric_difference(b)

    return d

d = set_ops({1,2,3},{3,4})
assert d["union"] == {1,2,3,4} and d["symdiff"] == {1,2,4}
````

````python
def dedupe_keep_first(xs: list):
    seen = set()
    liste = list()
    seen = set(xs)
    liste = list(seen)
    return liste

#dedupe_keep_first([1,2,1,3,2,4])
assert dedupe_keep_first([1,2,1,3,2,4]) == [1,2,3,4]
````

````python
def invert_index(items):
    """
    items: list of (item_id, tags_iterable).
    Return dict: tag -> set({item_id,...})
    """
    d = dict()
    for element in items:
        for letter in element[1]:
            if letter in d:
                d[letter].add(element[0])
            else:
                d[letter] = {element[0]}
    return d

    

ii = invert_index([(1,{"a","b"}),(2,{"b"}),(3,{"a"})])
assert ii == {"a":{1,3}, "b":{1,2}}
````

````python
def canonical_edge(a, b):
    """Return frozenset representing undirected edge."""
    return frozenset((a, b))
    
assert canonical_edge("A","B") == frozenset({"A","B"})
````

````python
def first_missing_positive(xs: list):
    """Return smallest missing positive integer."""
    set_list = set(xs)
    for number in range(1,len(xs)+1):
        if number > 0 and number not in set_list:
            return number

assert first_missing_positive([3,4,-1,1]) == 2
assert first_missing_positive([1,2,0]) == 3
````

````python
def has_triangle(edges: list):
    """
    edges is iterable of pairs (u,v). Return True if there exists
    a triangle u-v-w-u. Use sets for adjacency.
    """
    adj = {}

    for u, v in edges:
        if u not in adj:
            adj[u] = set()

        if v not in adj:
            adj[v] = set()

        adj[u].add(v)
        adj[v].add(u)

    for u, v in edges:
        common_neighbors = adj[u] & adj[v]
        if common_neighbors:
            return True

    return False


assert has_triangle([("a","b"),("b","c"),("c","a")]) is True
assert has_triangle([("a","b"),("b","c")]) is False
````
