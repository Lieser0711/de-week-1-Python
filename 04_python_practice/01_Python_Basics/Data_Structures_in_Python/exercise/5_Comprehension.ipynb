````python
def squares(n):
    return [num**2 for num in range(1, n+1)]

def odds(xs):
    return [elements for elements in xs if elements % 2 != 0]

assert squares(4) == [1,4,9,16]
assert odds([1,2,3,4]) == [1,3]
````

````python
def normalize_emails(items):
    """items: list of raw emails; return unique normalized set."""
    return set([elements.lower().strip() for elements in items])
assert normalize_emails([" A@X.com ","a@x.COM"]) == {"a@x.com"}
````

````python
def lengths_at_least(words, k=3):
    return {elements: len(elements) for elements in words if len(elements) >= k}

assert lengths_at_least(["a","bbb","cc"],3) == {"bbb":3}
````

````python
def transpose(m):
    length = len(m)
    num_elements = len(m[0])
    return [[m[idx][elements] for idx in range(length)] for elements in range(num_elements)]

assert transpose([[1,2,3],[4,5,6]]) == [[1,4],[2,5],[3,6]]
````

````python
def flatten_numbers(rows):
    """rows: mixed items; collect only ints/floats in one list."""
    return [idx for elements in rows for idx in elements if isinstance(idx, (int, float, complex)) and not isinstance(idx, bool)]

assert flatten_numbers([[1,"x"],[2.5,None],["y"]]) == [1,2.5]
````

````python
def freq_ge(xs, min_count=2):
    return {character: xs.count(character) for character in set(xs) if xs.count(character) >= min_count}

assert freq_ge("aabbbc",3) == {"b":3}
````

````python
def index_by_first_char(words):
    """Return dict: first_char -> list of words (keep order)."""
    return {elements[0]: [word for word in words if elements[0] == word[0]] for elements in set(words)}

assert index_by_first_char(["apple","art","bee","bat"]) == {"a":["apple","art"],"b":["bee","bat"]}
````

````python
def map_square_pos(xs):
    return [elements if elements <= 0 else elements**2 for elements in xs]

map_square_pos([-2,0,3]) == [-2,0,9]
````

````python
def two_letter_pairs(words):
    """Return set of first-two-letter tuples from alphabetic words only."""
    return {tuple(string[:2]) for string in words if string.isalpha() and len(string) >=2}
assert two_letter_pairs(["Hi!","oh","2fast","ok","bad","a"]) == {("o","h"), ("b","a"), ("o","k")}
````

````python
def grade_buckets(pairs):
    """
    pairs: iterable of (name, score 0..100).
    Buckets: A>=90, B>=80, C>=70, D>=60, F else.
    Return dict like {"A":[names...], "B":[...], ...} excluding empty buckets.
    """
    result = dict()
    for names, score in pairs:
        if score >= 90:
            grade = "A"
        elif score >= 80:
            grade = "B"
        elif score >= 70:
            grade = "C"
        elif score >= 60:
            grade = "D"
        else:
            grade = "F"
        if grade not in result:
            result[grade] = []
        result[grade].append(names)

    return{grade: names for grade, names in result.items() if names}

grades = grade_buckets([("Ana",95),("Bo",82),("Cy",67),("Di",50)])
assert grades["A"] == ["Ana"] and grades["B"] == ["Bo"] and grades["D"] == ["Cy"] and grades["F"] == ["Di"]
````
