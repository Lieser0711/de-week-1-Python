````python
def make_profile(name, email=None, country="IN"):
    result = {}
    result["name"] = name
    result["email"] = email
    result["country"] = country
    return result
p = make_profile("A","a@example.com")
assert p["name"]=="A" and p["country"]=="IN"
````

````python
def get_in(d, path, default=None):
    current = d
    for key in path:
        if key in current:
            current = current[key]
        else:
            return default
    return current


cfg = {"db":{"host":"localhost","port":5432}}
assert get_in(cfg, ["db","port"]) == 5432
assert get_in(cfg, ["db","user"], "postgres") == "postgres"
````

````python
def multimap_add(mm, key, value):
    """mm: dict key -> list; append value; create list if missing."""
    if key in mm:
        mm[key].append(value)
    else:
        mm[key] = [value]
    return mm

mm = {}
multimap_add(mm,"a",1); multimap_add(mm,"a",2)
assert mm == {"a":[1,2]}
````

````python
def invert_unique(d):
    base_dict = {}
    for key, value in d.items():
        if value in base_dict:
            continue
        else:
            base_dict[value] = key
    return base_dict

assert invert_unique({"a":1,"b":2}) == {1:"a",2:"b"}
````

````python
def invert_to_sets(d):
    base_dict= {}
    for key, value in d.items():
        if value in base_dict:
            base_dict[value].add(key)
        else:
            base_dict[value] = set(key)
    return base_dict
assert invert_to_sets({"a":1,"b":1,"c":2}) == {1:{"a","b"}, 2:{"c"}}
````

````python
def merge_priority(base, override):
    result = {**base, **override}
    return result
    
assert merge_priority({"a":1,"b":2},{"b":9,"c":3}) == {"a":1,"b":9,"c":3}
````

````python
def group_by(xs, key_fn):
    base_dict = {}
    for element in xs:
        current_key = key_fn(element)
        if current_key in base_dict:
            base_dict[current_key].append(element)
        else:
            base_dict[current_key] = [element]
    return base_dict

    
g = group_by(["apple","pear","plum","kiwi"], len)
assert g == {5:["apple"],4:["pear","plum","kiwi"]}
````

````python
def top_n(d, n):
    result = list(d.items())
    result.sort(reverse=True, key = lambda element: element[1])
    result = result[:n]
    return result
    

top_n({"a":3,"b":1,"c":5},2)
assert top_n({"a":3,"b":1,"c":5},2) == [("c",5),("a",3)]
````

````python
def left_join(left_rows, right_rows, key):
    """
    left_rows/right_rows: list of dicts
    Return: list of merged dicts; right fields may be missing.
    """
    result = []
    for elements in right_rows:
        for l_elements in left_rows:
            if elements["id"] == l_elements["id"]:
                right_keys = set(elements.keys())
                left_keys = set(l_elements.keys())
                add_keys = list(right_keys.difference(left_keys))
                for idx in add_keys:
                    l_elements[idx] = elements[idx]
                    result.append(l_elements)
            else:
                result.append(l_elements)
    return result




L = [{"id":1,"name":"A"},{"id":2,"name":"B"}]
R = [{"id":1,"score":90}]

out = left_join(L,R,"id")
assert out[0]["score"]==90 and out[1].get("score") is None
````

````python
def flatten_dot(d, parent_key=""):
    result = {}

    for key, value in d.items():

        if parent_key:
            new_key = parent_key + "." + key
        else:
            new_key = key

        if isinstance(value, dict):
            result.update(flatten_dot(value, new_key))
        else:
            result[new_key] = value

    return result

assert flatten_dot({"a":{"b":1},"c":2}) == {"a.b":1,"c":2}
````
