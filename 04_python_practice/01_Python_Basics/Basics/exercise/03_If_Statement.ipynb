1. **FizzBuzz (classic)**

   * `fizzbuzz(n)` → `"Fizz"`, `"Buzz"`, `"FizzBuzz"`, or `str(n)`.

   ```python
   def fizzbuzz(n):
       ...
   assert fizzbuzz(15) == "FizzBuzz"
   ```

````python
def fizzbuzz(n: int):
    if n%3 == 0 and n%5 == 0:
        print(n%3)
        print(n%5)
        return "FizzBuzz"
    elif n%3 == 0:
        return "Fizz"
    elif n%5 == 0:
        return "Buzz"
    else:
        return str(n)

fizzbuzz(9)
````


2. **Categorize temperature**

   * `temp_label(c)` → `freezing` (<0), `cold` (<10), `mild` (<20), `warm` (<30), `hot` (else).

   ```python
   def temp_label(c):
       ...
   ```


````python
def temp_label(c: int):
    if c < 0:
        return "freezing"
    elif c >= 0 and c < 10:
        return "cold"
    elif c >= 10 and c < 20:
        return "mild"
    elif c >= 20 and c < 30:
        return "warm"
    else:
        return "hot"

temp_label(-10)
````

3. **Truthy length**

   * `choose_label(s)` → `"empty"` if string is empty/whitespace; `"short"` (≤3), `"long"` (>3).

   ```python
   def choose_label(s):
       ...
   ```

````python
def choose_label(s):
    if not s or s.isspace():
        return "empty"
    elif len(s) <= 3:
        return "short"
    else:
        return "long"

choose_label("hallo")
````

4. **Grade boundaries**

   * `grade(score)` with 0–100 validation; raise `ValueError` otherwise. Return `A/B/C/D/F`.

   ```python
   def grade(score):
       ...
   ```

````python
def grade(score: int):
    if score < 0 or score > 100:
        raise ValueError("Value is lower than zero or bigger than 100!")
    elif score >= 0 and score < 10:
        return "F"
    elif score >= 10 and score < 20:
        return "D"
    else:
        return "A/B/C"

grade(100)
````


5. **Password strength (simple)**

   * `password_strength(s)` returns `"weak"`, `"medium"`, `"strong"` based on length and presence of digit/special.

    * "weak": less than 6 characters, or only letters or only digits

    * "medium": at least 6 characters, with both letters and digits

    * "strong": at least 8 characters, with letters, digits, and special characters

   ```python
   def password_strength(s):
       ...
   ```

````python
def password_strength(pw: str):
    if len(pw) < 6 or all(elements.isdigit() for elements in pw) or all(elements.isalpha() for elements in pw):
        return "weak"
    elif len(pw) >= 6 and any(elements.isdigit() for elements in pw) and any(elements.isalpha() for elements in pw) and all(not elements.isalnum() for elements in pw):
        return "medium"
    elif len(pw) >= 8 and any(elements.isdigit() for elements in pw) and any(elements.isalpha() for elements in pw) and any(elements.isalnum() for elements in pw):
        return "strong"

password_strength("hallo123%&/()")
````

6. **Shipping rules**

   * `shipping_cost(weight_kg, country, premium=False)`:

     * base by weight: ≤1kg: 5; ≤5kg: 10; else: 20
     * non-local country adds 15
     * premium adds +25 at the end

   ```python
   def shipping_cost(weight_kg, country, premium=False, local_country="IN"):
       ...
   ```

````python
def shipping_cost(weight_kg: int, country: str, premium=False, local_country="IN"):
    costs = 0
    if weight_kg <= 1:
        costs += 5
    elif weight_kg > 1 and weight_kg <= 5:
        costs += 10
    else:
        costs += 20 
    if country != local_country:
        costs += 15
    if premium:
        costs += 25
    return costs

shipping_cost(10, "ENG", premium=True)
````

7. **Pick earliest valid date**

   * Given three `(y,m,d)` tuples, `earliest_valid(dates)` where each must pass simple checks (1≤m≤12, 1≤d≤31) else ignore. Return earliest valid or `None`.

   ```python
   def earliest_valid(dates):
       ...
   ```

````python
def earliest_valid(dates: list):
    is_valid = 0
    for elements in dates:
        if elements[2] < 1 or elements[2] > 31 or elements[1] < 1 or elements[1] > 12:
            continue
        else:
            is_valid = 1
        if is_valid:
            return elements
        else:
            return None

test = earliest_valid([(2027, 10, 1), (2026, 30, 13), (2026,40,14)])
print(test)
````

8. **Safe cast**

   * `to_int(s, default=None)` returns int or default; only accept pure integer strings (e.g., `" 12 "` ok after strip, `"12.0"` not ok).

   ```python
   def to_int(s, default=None):
       ...
   ```

````python
def to_int(s: str, default = None):
    checked_s = s.strip()
    if any(letters == "." for letters in checked_s):
        return default
    else:
        return int(checked_s)

test = to_int(" 12.0 ")
print(test)
````

9. **Guard clauses**

   * `discount(price, code)` → if `price<=0` return 0; if `code=="NONE"` return `price`; if `code=="HALF"` return `price*0.5`; else return `price*0.9`.

   ```python
   def discount(price, code):
       ...
   ```

````python
def discount(price: int, code: str):
    if price <= 0:
        return 0
    elif code == "NONE":
        return price
    elif code == "HALF":
        return price * 0.5
    else:
        return price * 0.9
    
discount(100, "NONE")
````

10. **Ternary map**

    * `label_numbers(nums)` → for each n: `"neg" if n<0`, `"zero" if n==0`, `"pos" if n>0`. Use a list comprehension with a ternary chain internally (but focus on if-logic correctness).

    ```python
    def label_numbers(nums):
        ...
    assert label_numbers([-1,0,2]) == ["neg","zero","pos"]
    ```



````python
def label_numbers(nums: list):
    result = []
    for numbers in nums:
        result.append("neg" if numbers < 0 else "zero" if numbers == 0 else "pos")
    return result

label_numbers([-1,0,2])
````
