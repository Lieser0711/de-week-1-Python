### Warm-ups

1. **Middle char(s)**

   * `middle(s)` → If odd length return the middle char, else return the two middle chars.

   ```python
   def middle(s):
       ...
   assert middle("abc") == "b"
   assert middle("abba") == "bb"
   ```

````python
def middle(string: str):
    if len(string) % 2 == 0:
        return string[int(len(string)/2 -1) : int((len(string)/2 -1) * -1)]
    else:
        return string[int(round(len(string)/2) -1)]

middle('abbdcd')
````

2. **Title-case safely**

   * `safe_title(s)` → title case but do not lowercase words fully in ALL CAPS acronyms (e.g., `"learn SQL now"` → `"Learn SQL Now"`; `"use GPU"` → `"Use GPU"`).

   ```python
   def safe_title(s):
       ...
   ```

````python
def safe_title(string: str):
    list = string.split()
    new_format = []
    for words in list:
        if words.isupper():
            new_format.append(words)
        else:
            new_format.append(words.capitalize())
    return " ".join(new_format)
            
safe_title('learn SQL now')
````

3. **Reverse words**

   * `reverse_words(s)` → reverse word order but keep internal characters.

   ```python
   def reverse_words(s):
       ...
   assert reverse_words("one two  three") == "three two  one"
   ```

````python
def reverse_words(string: str):
    words= string.split()
    words.reverse()
    return " ".join(words)

reverse_words("one two three")
````

### Core

4. **Normalize spaces**

   * `squash_spaces(s)` → collapse multiple spaces/tabs to a single space; strip ends.

   ```python
   def squash_spaces(s):
       ...
   assert squash_spaces("  a\t b   c ") == "a b c"
   ```

````python
def squash_spaces(string: str):
    words = string.split()
    return " ".join(words)

squash_spaces("  a\t b   c ")
````

5. **CSV line → list (no quotes)**

   * `split_csv(s)` → split by commas, trim whitespace around fields; ignore empty trailing field if line ends with comma.

   ```python
   def split_csv(s):
       ...
   assert split_csv(" a, b ,c,") == ["a","b","c"]
   ```
  

````python
def split_csv(string:str):
    liste = []
    liste = string.replace(",", " ").split()
    return liste

split_csv(" a, b ,c,") 
````

6. **Mask secrets**

   * `mask_email(s)` → keep first char of user and domain, mask the rest with `*`, keep TLD.

     * `"alice@example.com"` → `"a****@e******.com"`

   ```python
   def mask_email(s):
       ...
   ```

````python
def mask_email(email):
    name, domain = email.split('@')
    domain_name, extension = domain.rsplit('.', 1)

    return (
        name[0] + '*' * (len(name) - 1)
        + '@'
        + domain_name[0] + '*' * (len(domain_name) - 1)
        + '.' + extension
    )

mask_email('liesa@example.com')
````

7. **Find all indexes**

   * `find_all(s, sub)` → list of start indices where `sub` occurs (including overlaps).

   ```python
   def find_all(s, sub):
       ...
   assert find_all("aaaa", "aa") == [0,1,2]
   ```

````python
def find_all(s:str, sub:str):
    result = []

    for i in range(len(s) - len(sub) + 1):
        if s[i:i + len(sub)] == sub:
            result.append(i)

    return result
````

8. **Anagram check**

   * `is_anagram(a, b)` ignoring spaces, case, punctuation.[Anagram](https://en.wikipedia.org/wiki/Anagram)

   ```python
   def is_anagram(a, b):
       ...
   assert is_anagram("Listen", "Silent")
   ```

````python
def is_anagram(a: str, b: str):
    return sorted(a.lower()) == sorted(b.lower())

is_anagram("Listen", "Silent")
````

9. **Format table row**

   * `fmt_row(values, widths)` → left-align strings to fixed widths, joined by `" | "`.

   ```python
   def fmt_row(values, widths):
       ...
   assert fmt_row(["a","bb"], [3,4]) == "a   | bb  "
   ```

````python
def fmt_row(values, widths):
    result = []

    for value, width in zip(values, widths):
        result.append(value.ljust(width))

    return " | ".join(result)
````
